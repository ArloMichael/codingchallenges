```python
import matplotlib.pyplot as plt
from random import randint
```

```python
months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun", "Jul", "Aug", "Sep", "Oct", "Nov", "Dec"]
temps = [randint(1000, 5000) for _ in months]
colors = ['tab:red', 'tab:blue', 'tab:red', 'tab:orange'] # or 'colours' idk i think its convention to use 'color'
```

```python
plt.bar(months, temps, color=colors)

plt.xlabel("Month")
plt.ylabel("Sales")
plt.title("Monthly sales data")

plt.show()
```
