```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
data = np.clip(np.random.normal(loc=50, scale=16.67, size=100), 0, 100)
```

```python
counts, bins, patches = plt.hist(data, bins=10)

cmap = plt.cm.viridis

cent = 0.5 * (bins[:-1] + bins[1:])
norm = (cent - cent.min()) / (cent.max() - cent.min())

for center, patch in zip(norm, patches):
    color = cmap(center)
    patch.set_facecolor(color)

plt.xlabel("Score")
plt.ylabel("Students")
plt.title("Exam scores")

plt.show()
```
