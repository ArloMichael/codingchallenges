```python
import matplotlib.pyplot as plt
from random import randint
```

```python
browser_names = ["Chrome", "Firefox", "Safari", "Edge"]
browser_shares = [randint(0, 100) for _ in browser_names]
colors = ['tab:red', 'tab:blue', 'tab:red', 'tab:orange']
```

```python
plt.pie(browser_shares, labels=browser_names, autopct='%1.1f%%')

plt.title("Web Browser market share")

plt.show()
```
