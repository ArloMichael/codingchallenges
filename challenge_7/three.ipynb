```python
import matplotlib.pyplot as plt
from random import randint, gauss
```

```python
n = 500
male_height = [randint(150, 200) for _ in range(n)]
male_weight = [
    round(max(17, min(35, gauss(24, 3))) * (height / 100) ** 2)
    for height in male_height
]
female_height = [randint(145, 190) for _ in range(n)]
female_weight = [
    round(max(16, min(35, gauss(23, 3))) * (height / 100) ** 2)
    for height in female_height
]
```

```python
plt.scatter(male_height, male_weight)
plt.scatter(female_height, female_weight)
plt.xlabel("Height")
plt.ylabel("Weight")
plt.title("Height and Weight graph")

plt.show()
```
